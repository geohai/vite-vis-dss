# Convert JSON to GeoParquet

```python
# load libraries
import geopandas as gpd
```

```python
# read in the JSON file
gdf = gpd.read_file('/Users/ilyonsg/Documents/nrel/vis-dss/data/json/P13U.json')
```

```python
# keep just the points
buses = gdf[gdf['geometry'].type == 'Point']

# keep just the lines
lines = gdf[gdf['geometry'].type == 'LineString']
```

```python
# export to geoparquet
buses.to_parquet(
    path='/Users/ilyonsg/Documents/nrel/vis-dss/data/parquet/P13Ubuses.parquet', 
    index=False, 
    compression='BROTLI',
    geometry_encoding='geoarrow' # geoarrow doesn't support multiple types
)
```

```python
lines.to_parquet(
    path='/Users/ilyonsg/Documents/nrel/vis-dss/data/parquet/P13Ulines.parquet', 
    index=False, 
    compression='BROTLI',
    geometry_encoding='geoarrow' # geoarrow doesn't support multiple types
)
```
